```python
import numpy as np
from sympy import Matrix

d1 = 5
d2 = 1

A = np.array([
    [d1 + 2, d2 + 1],
    [2 * d1, d2 + 2]
], dtype=float)

print("Matrix A:\n", A)
shape_A = A.shape
det_A = np.linalg.det(A)
rank_A = np.linalg.matrix_rank(A)
eigvals_A, eigvecs_A = np.linalg.eig(A)
inv_A = np.linalg.inv(A) if det_A != 0 else None

print("Shape A:", shape_A)
print("Det A:", det_A)
print("Rank A:", rank_A)
print("Eigenvalues A:", eigvals_A)
print("Inverse A:\n", inv_A)
B = A.copy()
B[0, 0] += 1

print("\nMatrix B:\n", B)
shape_B = B.shape
det_B = np.linalg.det(B)
rank_B = np.linalg.matrix_rank(B)
eigvals_B, eigvecs_B = np.linalg.eig(B)
inv_B = np.linalg.inv(B) if det_B != 0 else None

print("Shape B:", shape_B)
print("Det B:", det_B)
print("Rank B:", rank_B)
print("Eigenvalues B:", eigvals_B)
print("Inverse B:\n", inv_B)
```

5. Comparison & Explanation
1.	How did the determinant change and why?
The determinant increased from 1 to 4 (a net change of +3). For a 2 × 2 matrix, det = ad - bc. Adding +1 to element a changes the determinant to (a + 1)d - bc = (ad - bc) + d. Because element d = 3, the determinant increased by exactly +3.
2.	Did the rank change? Explain.
No, the rank remained 2. Both matrix A and matrix B have non-zero determinants (det(A) = 1, det(B) = 4), which guarantees that all rows and columns in both matrices remain linearly independent.
3.	How did the eigenvalues respond to the value change?
Both eigenvalues shifted to higher positive values. The trace increased from 10 to 11, and the trace equals the sum of the eigenvalues. Because both the sum and the product increased, both eigenvalues shifted upward.
4.	Is B easier or harder to invert than A? Why?
Matrix B is computationally easier to invert. Matrix B has a determinant further away from zero and a lower condition number. A lower condition number makes matrix inversion far less sensitive to numerical rounding errors.
