```python
import numpy as np
import matplotlib.pyplot as plt

#Data Generation
regular = np.random.uniform(2, 12, 3000)
busy = np.clip(np.random.normal(8, 2, 3000), 0, None)

#Visualization (Overlaid Histograms)
plt.hist(regular, bins=30, alpha=0.5, label='Regular Day', color='gray',edgecolor='black')
plt.hist(busy, bins=30, alpha=0.5, label='Busy Day', color='pink',edgecolor='black')
plt.xlabel('Wait Time (minutes)')
plt.ylabel('Frequency')
plt.title('Customer Wait Times')
plt.show()

# Statistical Analysis
for name, data in [("Regular Day", regular), ("Busy Day", busy)]:
    print(f"{name}:")
    print(f"  Mean:   {np.mean(data):.2f}")
    print(f"  Median: {np.median(data):.2f}")
    print(f"  Std:    {np.std(data):.2f}")
    print(f"  Max:    {np.max(data):.2f}\n")

```
