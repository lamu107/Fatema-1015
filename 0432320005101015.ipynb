```python
import math, random
import numpy as np

R = 15
r = R / 10.0
base_points = 200 * 5
print(
    f"{'Range Used':<12} | {'Total Points':<12} | {'Points Inside':<14} | {'Estimated Pi':<12} | {'Error':<10}"
)
print("-" * 70)
for i in range(1, 6):
    pts = base_points * i
    half = pts // 2
    in_rand = sum(
        random.uniform(0, r) ** 2 + random.uniform(0, r) ** 2 <= r**2
        for _ in range(half)
    )
    in_np = np.sum(
        (np.random.rand(pts - half) * r) ** 2
        + (np.random.rand(pts - half) * r) ** 2
        <= r**2
    )
    inside = in_rand + in_np
    pi_est = 4 * inside / pts
    error = abs(math.pi - pi_est)
    print(
        f"[0, {r:.1f}]      | {pts:<12} | {inside:<14} | {pi_est:<12.5f} | {error:<10.5f}"
    )
```
